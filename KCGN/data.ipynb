```python
data_dir = os.path.join('dataset', 'ciao_timestamp')
rating = loadmat(os.path.join(data_dir, 'rating_with_timestamp.mat'))
trust = loadmat(os.path.join(data_dir, 'trust.mat'))
```

```python
rating = rating['rating'].astype(np.int32)
rating_df = pd.DataFrame(rating, columns=['user_id', 'product_id', 'category_id', 'rating', 'helpfulness', 'timestamp'])
rating_df = rating_df[['user_id','product_id','category_id','rating','timestamp']]
```

```python
trust = trust['trust'].astype(np.int32)
trust_df = pd.DataFrame(trust, columns=['user_id_1','user_id_2'])
```

```python
rating_df.to_csv(os.path.join(data_dir, 'rating_timestamp_org.csv'))
trust_df.to_csv(os.path.join(data_dir, 'trustnetwork_org.csv'))
```

```python
print(rating_df.shape, trust_df.shape)
```

```python
# data preprocessing (same /w SoFT)

# 1. filter data
rating_df = rating_df.dropna(how='any')
rating_df = rating_df.drop_duplicates(['user_id','product_id'], keep='first')
rating_df = rating_df[rating_df['rating'].between(1,5,'both')]
trust_df = trust_df.dropna(how='any')
trust_df = trust_df.drop_duplicates(keep='first')

# 2. encoding(user_id, product_id)
total_users = set(rating_df.user_id.unique()).union(set(trust_df.user_id_1.unique())).union(set(trust_df.user_id_2.unique()))
user_dict = dict(zip(total_users, range(len(total_users))))
total_items = rating_df.product_id.unique()
item_dict = dict(zip(total_items, range(len(total_items))))
rating_df['user_id'] = rating_df['user_id'].map(user_dict)
rating_df['product_id'] = rating_df['product_id'].map(item_dict)
trust_df['user_id_1'] = trust_df['user_id_1'].map(user_dict)
trust_df['user_id_2'] = trust_df['user_id_2'].map(user_dict)
rating_df['category_id'] = rating_df['category_id']-rating_df['category_id'].min()

# # 3. shuffle and split data
# shuffled_rating = shuffle(rating_df, random_state=42)
# num_test = int(len(shuffled_rating)*0.2)
# rating_test = shuffled_rating.iloc[:num_test//2]
# rating_valid = shuffled_rating.iloc[num_test//2:num_test]
# rating_train = shuffled_rating.iloc[num_test:]
```

```python
# rating mat / time mat
u = rating_df['user_id']
p = rating_df['product_id']
r = rating_df['rating']
t = rating_df['timestamp']
rating_mat = csr_matrix((r, (u,p)))
time_mat = csr_matrix((t, (u,p)))
```

```python
def buildSubGraph(mat):
    # build sub graph
    subNode = 10
    node_num = mat.shape[0]
    graph = nx.Graph(mat)
    subGraphList = list(nx.connected_components(graph)) # sub graph list반환
    subGraphCount = len(subGraphList)
    node_subGraph = [-1 for _ in range(node_num)]
    adjMat = sparse.dok_matrix((subGraphCount, node_num), dtype=np.int)
    node_list = []
    for i in range(len(subGraphList)):
        subGraphID = i
        subGraph = subGraphList[i]
        if len(subGraph)>subNode: # sub graph 노드의 개수가 10개보다 많은 경우
            node_list += list(subGraph)
        for node_id in subGraph:
            assert node_subGraph[node_id] == -1
            node_subGraph[node_id] = subGraphID # node_subGraph의 값을 update(sub graph의 index로 업데이트)
            adjMat[subGraphID, node_id] = 1 # 해당 노드가 어떤 sub graph에 속해있는지에 대한 adj matrix
    node_subGraph = np.array(node_subGraph)
    assert np.sum(node_subGraph==-1)==0
    adjMat = adjMat.tocsr()
    return subGraphList, node_subGraph, adjMat, node_list

_, uu_node_subGraph, uu_subGraph_adj, uu_dgi_node = buildSubGraph(uuMat)
_, ii_node_subGraph, ii_subGraph_adj, ii_dgi_node = buildSubGraph(iiMat)
```

```python
# sparse matrix(subGraph_adj : sub graph에 대한 adj matrix)
def sparse_mx_to_tensor(mx):
    mx = mx.tocoo().astype(np.float32)
    indices = torch.from_numpy(np.vstack((mx.row, mx.col)).astype(np.int64))
    values = torch.from_numpy(mx.data)
    shape = torch.Size(mx.shape)
    return torch.sparse.FloatTensor(indices, values, shape)

uu_subGraph_adj_tensor = sparse_mx_to_tensor(uu_subGraph_adj)
uu_subGraph_adj_norm = torch.from_numpy(np.sum(uu_subGraph_adj, axis=1)).float()
ii_subGraph_adj_tensor = sparse_mx_to_tensor(ii_subGraph_adj)
ii_subGraph_adj_norm = torch.from_numpy(np.sum(ii_subGraph_adj, axis=1)).float()
```
