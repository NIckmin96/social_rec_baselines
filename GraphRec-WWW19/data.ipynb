```python
import os
import pickle
import pandas as pd
import numpy as np
import scipy.sparse as sparse
```

# spd matrix

```python
spd_matrix = sparse.load_npz(os.path.join('data','ciao_timestamp','shortest_path_result.npz'))
```

```python
def get_col_indices_by_row(mtx, target_val):
    # 1. 전체 data에서 target_val과 일치하는 마스크 생성
    mask = (mtx.data == target_val)
    
    # 2. 마스크를 적용해 조건에 맞는 column indices만 추출
    filtered_cols = mtx.indices[mask]
    
    # 3. 각 행(row)별로 조건에 맞는 데이터가 몇 개 있는지 계산
    # np.add.reduceat을 사용하여 indptr이 가리키는 행 구간별로 mask(True=1)를 합산합니다.
    row_counts = np.add.reduceat(mask, mtx.indptr[:-1])
    
    # 4. row_counts를 기준으로 filtered_cols를 분할하여 행별 리스트 생성
    split_indices = np.cumsum(row_counts)[:-1]
    result_per_row = np.split(filtered_cols, split_indices)
    
    return result_per_row

# 결과 실행
result = get_col_indices_by_row(spd_matrix, 2)
```
