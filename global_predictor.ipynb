```python
import os
import pandas as pd
import numpy as np
```

```python
# dataset dir
ciao_dir = os.path.join('SocialMF','data','ciao_timestamp')
epi_dir = os.path.join('SocialMF','data','epinions')
yelp_dir = os.path.join('SocialMF','data','yelp')
```

```python
# load ciao
ciao_all = pd.read_csv(os.path.join(ciao_dir, 'rating_org.csv'))
ciao_train = pd.read_csv(os.path.join(ciao_dir, 'rating_train_42.csv'))
ciao_test = pd.read_csv(os.path.join(ciao_dir, 'rating_test_42.csv'))
```

```python
def global_predictor(train, test):
    mean = train.rating.mean()
    diff = test.rating.values - mean
    mse = np.sum(diff**2)/len(test)
    rmse = np.sqrt(mse)
    mae = np.sum(np.abs(diff))/len(test)
    
    return rmse, mae
```

```python
# ciao global predictor
rmse, mae = global_predictor(ciao_train, ciao_test)
print(f"Ciao RMSE : {rmse} / MAE : {mae}")
```

```python
# load epinions
epi_all = pd.read_csv(os.path.join(epi_dir, 'rating_org.csv'))
epi_train = pd.read_csv(os.path.join(epi_dir, 'rating_train_42.csv'))
epi_test = pd.read_csv(os.path.join(epi_dir, 'rating_test_42.csv'))
```

```python
# epinions global predictor
rmse, mae = global_predictor(epi_train, epi_test)
print(f"Epinions RMSE : {rmse} / MAE : {mae}")
```

```python
# load yelp
yelp_all = pd.read_csv(os.path.join(yelp_dir, 'rating_org.csv'))
yelp_train = pd.read_csv(os.path.join(yelp_dir, 'rating_train_42.csv'))
yelp_test = pd.read_csv(os.path.join(yelp_dir, 'rating_test_42.csv'))

# epinions global predictor
rmse, mae = global_predictor(yelp_train, yelp_test)
print(f"Yelp RMSE : {rmse} / MAE : {mae}")
```
