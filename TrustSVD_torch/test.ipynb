```python
import os
import numpy as np
import scipy.sparse as sparse
```

```python
trust_mat = sparse.load_npz(os.path.join('data','ciao_timestamp','trust_matrix.npz'))
```

```python
user_user = [trust_mat.indices[trust_mat.indptr[i]:trust_mat.indptr[i+1]] for i in range(trust_mat.shape[0])]
```

```python
user_user_len = list(map(lambda x:len(x), user_user))
```

```python
trust_mat = trust_mat.todok()
user_user_dict = {}
for u in range(trust_mat.shape[0]):
    user_user_dict[u] = list(map(lambda x:x[1], trust_mat.todok()[u].keys()))
```
